```python
n = int(input("Digite o número de Itens aqui: "))
k = int(input("Digite o número de pessoas aqui: "))

itens_por_pessoa = n // k
itens_restantes = n % k

print("Itens por pessoa:", itens_por_pessoa)
print("Itens restantes:", itens_restantes)
```

```python
altura = int(input("Digite a altura do Retângulo aqui: "))
base = int(input("Digite a base do Retângulo aqui: "))

area = altura * base
perimetro = 2 * (altura + base)

print("Área do Retângulo:", area)
print("Perímetro do Retângulo:", perimetro)
```

```python
a = int(input("Digite o valor de \"a\" aqui: "))
b = int(input("Digite o valor de \"b\" aqui: "))
c = int(input("Digite o valor de \"c\" aqui: "))

e1 = a + b*c
e2 = (a + b)*c
e3 = a**2 + b**2 + c**2

print(f"E1 = {e1}")
print(f"E2 = {e2}")
print(f"E3 = {e3}")

```

```python
c = float(input("Digite a temperaturas  em Celcius aqui: "))

fahrenheit = 1.8*c + 32

f = float(input("Digite a temperatura em Fahrenheit aqui: "))

celcius = (f - 32) / 1.8

print(f"{c} Celcius = {fahrenheit} Fahrenheit")
print(f"{f} Fahrenheit = {celcius} Celcius")
```

```python
n = int(input("Digite o valor de N"))

print(f"N é par? {n % 2 == 0}" ) # Retorna True se "n" for par e False se "n" for ímpar
print(f"N é divisível por 3? {n % 3 == 0}") # Retorna True se "n" for divisível por 3 e False se "n" não for divisível por 3
print(f"N é divisível por 6? {(n % 2 == 0) and (n % 3 == 0)}") # Retorna True se "n" for divisível por 6 e False se "n" não for divisível por 6
```

```python
nota1 = float(input("Digite sua primeira nota aqui: "))
peso1 = float(input("Digite o peso da primeira nota aqui: "))

nota2 = float(input("Digite sua segunda nota aqui: "))
peso2 = float(input("Digite o peso da segunda nota aqui: "))

nota3 = float(input("Digite sua terceira nota aqui: "))
peso3 = float(input("Digite o peso da terceira nota aqui: "))

media_ponderada = ((nota1 * peso1) + (nota2 + peso2) + (nota3 + peso3)) / (peso1 + peso2 + peso3) 

print(f"Sua média é: {media_ponderada}")


```
